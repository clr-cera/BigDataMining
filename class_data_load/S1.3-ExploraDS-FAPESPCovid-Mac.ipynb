 ## A tabela `Desfecho` permite tratar a evolução __histórica__.<br>

Vamos avaliar os vários atendimentos de cada paciente,<br>
&emsp; escolher alguns poucos pacientes para _entender_ como isso é feito, usando, por exemplo, os pacientes de `Guarulhos` (porque tem poucos pacientes),<br>
&emsp; e listar cada atendimento, agregando:
 * um atributo com a `ordem` do atendimento de cada paciente,
 * um atributo com o `separação` em número de dias __entre esse atendimento e o atendimento anterior__,
 * um atributo com o `tempo de atendimento`, também em número de dias __entre esse atendimento e o primeiro atendimento registrado desse paciente__.

```python
%%sql
SELECT P.id_paciente,
       '## '||RoW_Number(*)OVER(PARTITION BY P.id_paciente ORDER BY D.dt_atendimento)||' ##' Ordem, 
       D.dt_atendimento, D.de_clinica, D.de_desfecho,
       D.dt_atendimento -  lag(D.dt_atendimento) OVER (PARTITION BY P.id_paciente ORDER BY D.dt_atendimento) Separacao,
       D.dt_atendimento -  FIRST_Value(D.dt_atendimento) OVER (PARTITION BY P.id_paciente ORDER BY D.dt_atendimento) Tempo_Atend
    FROM Pacientes P JOIN Desfechos D ON P.ID_Paciente = D.ID_Paciente
    WHERE CD_Municipio ='GUARULHOS'
    ORDER by P.id_paciente, D.dt_atendimento
    LIMIT 20;
```

Sabendo disso:<br>

<font color="#006064" face="Georgia" style="border: 2px solid #4DD0E1; padding: 0px; background-color:#B2EBF2; line-height: 1.0;" size=4>
<i>Como podemos obter o desfecho de cada paciente</i>?</font><br><br>

Vamos considerar o atendimento com a data mais recente como sendo o estado _conhecido_ do paciente.

Portanto:<br>
<font  size=3>&#128073;&#127997;</font>
para estabelecer se um paciente veio a óbito, é necessário:
 * fazer uma JUNÇÃO da tabela `Pacientes` 
 * com uma relação que indica __<font color='red'>qual é o desfecho mais recente</font>__ de cada paciente

```python
%%sql
SELECT P.*, FD.De_FDesfecho, FD.Dt_FAtendimento
    FROM Pacientes P LEFT JOIN
        (SELECT  ID_Paciente, Dt_FAtendimento, De_FDesfecho
             FROM (SELECT ID_Paciente,
                       MAX(Dt_Atendimento) OVER(Partition BY ID_Paciente) AS Dt_FAtendimento, 
                       MAX(De_desfecho) OVER(Partition BY ID_Paciente) AS De_FDesfecho
                   FROM Desfechos
                   ) As temp
             GROUP BY ID_Paciente, Dt_FAtendimento, De_FDesfecho
        ) AS FD -- Desfecho Final 
            ON P.id_paciente=FD.ID_paciente
    LIMIT 10;
```

<br>Usando essa tabela de pacientes, podemos verificar como está a distribuição das idades, mas considerando agora apenas os __<font color='red'> pacientes com desfecho em óbito</font>__.

```python
ObitosIdade = pdsql.read_sql(text("""
SELECT Bins.B AS Idade,\
          CASE WHEN Tab.Conta IS NULL THEN 0 ELSE Tab.Conta END Tot
    FROM (WITH Lim AS (SELECT Max(2021-AA_Nascimento) Ma FROM D2.Pacientes)
                       SELECT Generate_Series(1, Lim.Ma) AS B FROM Lim) AS Bins
           LEFT OUTER JOIN
       (SELECT 2021-AA_Nascimento AS Idade, Count(*) Conta
           FROM D2.Pacientes P LEFT JOIN
               (SELECT  ID_Paciente, Dt_FAtendimento, De_FDesfecho
                    FROM (SELECT ID_Paciente,
                                 MAX(Dt_Atendimento) OVER(Partition BY ID_Paciente) AS Dt_FAtendimento, 
                                 MAX(De_desfecho) OVER(Partition BY ID_Paciente) AS De_FDesfecho
                              FROM D2.Desfechos
                          )	 As temp
                    GROUP BY ID_Paciente, Dt_FAtendimento, De_FDesfecho
                ) AS FD
                     ON P.id_paciente=FD.ID_paciente
           WHERE FD.DE_FDesfecho~*'[oó]bito'
           GROUP BY 1
        ) AS Tab ON Bins.B=Tab.Idade
    ORDER BY 1;"""), engine)

ObitosIdade.plot(x="idade", y="tot", kind="bar", figsize=(20, 5))
```

De fato... Houve mesmo uma tendência de haver mais óbitos em pacientes com idades mais avançadas...

Qual a proporção de óbitos?

```python
tt=pdsql.read_sql(text("""
SELECT Count(*) todos, Count (*) FILTER (WHERE FD.DE_FDesfecho~*'[oó]bito') obitos
    FROM D2.Pacientes P LEFT JOIN
        (SELECT  ID_Paciente, Dt_FAtendimento, De_FDesfecho
            FROM (SELECT ID_Paciente,
                                 MAX(Dt_Atendimento) OVER(Partition BY ID_Paciente) AS Dt_FAtendimento, 
                                 MAX(De_desfecho) OVER(Partition BY ID_Paciente) AS De_FDesfecho
                              FROM D2.Desfechos
                 ) As temp
            GROUP BY ID_Paciente, Dt_FAtendimento, De_FDesfecho
        ) AS FD  ON P.id_paciente=FD.ID_paciente"""
    ), engine)

display(HTML(f"\n<b>Total de pacientes: </b>{tt["todos"][0]} &emsp; <b>Óbitos: </b>{tt["obitos"][0]} &emsp; <b>Porcentagem de óbitos: </b>{ (100.*tt["obitos"][0])/tt["todos"][0]}"))
```

## 6. Explorar a tabela `ExamLabs`

A tabela `ExamLabs`  armazena os resultados dos `Exames Laboratoriais` a que os pacientes foram submetidos.\
&#x26A0; Essa tabela é maiorzinha, a consulta toma algum tempo!

Vamos olhar essa tabela:

```python
%%sql
SELECT *
    FROM ExamLabs
    ORDER BY Random()
    LIMIT 5;
```

 * Cada `Paciente` (identificado pelo seu `Id_Paciente`) é atendido no contexto de um `Atendimento` (identificado pelo seu `Id_Atendimento`) que pode durar um ou mais dias.\
 * Cada  `Exame` é realizado numa data (`Dt_Coleta`) por solicitação de algum departamento onde o paciente foi atendido (`De_Origem`).
 * Cada  `Exame` tem um tipo (`DE_Exame`) e pode medir um ou mais analitos (`De_Analito`).
 * A medida tem um valor (`De-Resultado`) medido numa unidade (`CD_Unidade`), e pode ter um valor de referência (`CD_ValorReferencia`).
 * Para poder armazenar qualquer tipo de valor, o valor desse atributo é armazenado como um texto.<br>
   O atributo `DE_ResultNum` tem o valor como um tipo numérico para dados que são numéricos.

<table style="margin-left: 0; width: 800px;">
    <tr><td style="border: 2px solid #4DD0E1; background-color: #B2EBF2;">
    <font size=4>&#128073;&#127997;</font> <font size=3> Assume-se que um mesmo exame não se repete no mesmo dia.<br><br>
    &emsp; Mas isso não é verdade, pois existem ocorrências de mais de um exame do mesmo paciente no mesmo dia e atendimento.<br>
    &emsp; Quando isso ocorre, nos exames que avaliam mais de um analito, não é possível saber quais medidas compõem cada exame.
    Nesse caso, quando for necessário usar um valor, iremos escolher a alternativa de obter a média de todos os valores do mesmo atributo daquele dia.</font>
    </td></tr></table>

Para obter todos os dados de um exame completo, devem ser indicados os atributos:
   `Id_Paciente`, `Id_Atendimento`, `Dt_Coleta` e `De_Exame`.
Por exemplo:

```python
%%sql
SELECT De_Analito, De_Resultado, Cd_Unidade, De_ValorReferencia
    FROM ExamLabs
    WHERE Id_Paciente='DD4A17C40AE74BBAD93508DFAC55A4F9' AND
          Id_Atendimento='0FD2D8CD180DF69D73DCCBA558EC0138' AND
          Dt_Coleta='2021-03-14' AND
          De_Exame ~* 'Hemograma'
```
